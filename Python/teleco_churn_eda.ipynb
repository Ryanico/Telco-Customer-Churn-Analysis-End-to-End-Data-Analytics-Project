```python
#Which contract type has the highest churn rate?
df.groupby('Contract')["Churn"].value_counts()
```

```python
contract_type_churn_rate = (df.groupby('Contract')["Churn"].value_counts(normalize=True).unstack().mul(100))
```

```python
contract_type_churn_rate['Yes'].sort_values(ascending=False)
```

## The contract with the highest churn rate is a month-to-month contract
## The contract with the lowest churn rate is a Two year contract

```python
# Does internet service type influence churn?
internet_service_type_churn_rate = (df.groupby('InternetService')["Churn"].value_counts(normalize=True).unstack().mul(100))
```

```python
internet_service_type_churn_rate["Yes"].sort_values(ascending=False)
```

# Yes. Internet services type influences churn rate. The internet service type with the highest churn rate is Fiber optic

# Comparing Churn Rates across all services

```python
service_columns = [
    "PhoneService",
    "MultipleLines",
    "InternetService",
    "OnlineSecurity",
    "OnlineBackup",
    "DeviceProtection",
    "TechSupport",
    "StreamingTV",
    "StreamingMovies"
]
```

```python
for service in service_columns:
    churn_rate = (
        df.groupby(service)["Churn"]
          .apply(lambda x: (x == "Yes").mean() * 100)
          .round(2)
    )
    
    print(f"\n{service}")
    print(churn_rate)
```

## InternetService has the highest observed churn, with Fiber optic customers experiencing a churn rate of approximately 42%. However, OnlineSecurity and TechSupport also show similarly high churn rates among customers who do not subscribe to these services. This indicates that Fiber optic customers and customers without additional support and security services may be important segments to investigate for customer retention strategies.

```python
# Do customers with higher monthly charges churn more?
monthly_charges_churn_rate = (df.groupby('MonthlyCharges')["Churn"].value_counts(normalize=True).unstack().mul(100))
```

```python
monthly_charges_churn_rate["Yes"].sort_values(ascending=False).head(10)
```

# The monthly charges churn rate varies a lot and not only the ones with the highest monthly charges have churned more, but also the ones with low monthly charges have high churn rates as well.

```python
# Do customers with higher total charges churn more?
total_charges_churn_rate = (df.groupby('TotalCharges')["Churn"].value_counts(normalize=True).unstack().mul(100))
```

```python
total_charges_churn_rate["Yes"].sort_values(ascending=False).head(10)
```

# Some of the top highest total charges have 100% churn rate. My conclusion is that higher total charges have higher churn rates

```python
# Is there a relationship between tenure and total charges?
corr = df.corr(numeric_only=True)
corr
```

```python
sns.heatmap(corr, annot = True)

plt.rcParams['figure.figsize'] = (20,7)

plt.show()
```

| Correlation | Meaning                        |
| ----------: | ------------------------------ |
|         1.0 | Perfect positive relationship  |
|         0.8 | Strong positive relationship   |
|         0.5 | Moderate positive relationship |
|           0 | No linear relationship         |
|        -0.5 | Moderate negative relationship |
|        -0.8 | Strong negative relationship   |
|        -1.0 | Perfect negative relationship  |

# The relationship between tenure and total charges is a strong positive relationship according to the correlation heatmap.
